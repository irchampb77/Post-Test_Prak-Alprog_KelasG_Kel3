```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Mempercantik grafik
sns.set_style("whitegrid")

print("Library berhasil diimport")
```

```python
df = pd.read_csv("Kelas G_Fixed Call Report.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df.columns.tolist())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
```

```python
max_service = (
    df.groupby('Year')
    ['Number of Calls To Know About The Service']
    .max()
    .reset_index()
)

max_service
```

```python
fig, ax = plt.subplots(figsize=(10,5))

sns.barplot(
    data=max_service,
    x='Year',
    y='Number of Calls To Know About The Service',
    ax=ax
)

ax.set_title('Nilai Maksimum Panggilan Informasi Layanan per Tahun')
ax.set_xlabel('Tahun')
ax.set_ylabel('Jumlah Panggilan')

plt.show()
```
