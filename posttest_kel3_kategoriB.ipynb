Perintah: tampilkan visualisasi tren fluktuasi bulanan angka konsultasi medis dokter (`Total Number of Doctors Consultancy`) khusus sepanjang tahun 2022 dalam bentuk Line Chart.

```python
import csv
import matplotlib.pyplot as plt
from pathlib import Path

plt.style.use("ggplot")
```

```python
file_path = Path("Post-Test_Prak-Alprog_KelasG_Kel3") / "Kelas G_Fixed Call Report.csv"

if not file_path.exists():
    file_path = Path("Kelas G_Fixed Call Report.csv")

with file_path.open(newline="", encoding="utf-8-sig") as file:
    rows = list(csv.DictReader(file))

rows[:5]
```

```python
month_order = {
    "Jan": 1,
    "Feb": 2,
    "Mar": 3,
    "Apr": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "Aug": 8,
    "Sep": 9,
    "Sept": 9,
    "Oct": 10,
    "Nov": 11,
    "Dec": 12,
}

data_2022 = []

for row in rows:
    if int(row["Year"]) == 2022:
        data_2022.append({
            "Year": int(row["Year"]),
            "Month": row["Month"],
            "Total Number of Doctors Consultancy": int(row["Total Number of Doctors Consultancy"]),
            "Month_Order": month_order[row["Month"]],
        })

data_2022 = sorted(data_2022, key=lambda row: row["Month_Order"])

print("Year  Month  Total Number of Doctors Consultancy")
for row in data_2022:
    print(f"{row['Year']}  {row['Month']:<5}  {row['Total Number of Doctors Consultancy']}")
```

```python
months = [row["Month"] for row in data_2022]
doctor_consultancy = [row["Total Number of Doctors Consultancy"] for row in data_2022]

plt.figure(figsize=(12, 6))

plt.plot(
    months,
    doctor_consultancy,
    marker="o",
    linewidth=2.5,
    color="#1f77b4",
)

for month, value in zip(months, doctor_consultancy):
    plt.text(month, value, f"{value:,}", ha="center", va="bottom", fontsize=8)

plt.title("Tren Bulanan Konsultasi Dokter Tahun 2022", fontsize=14, fontweight="bold")
plt.xlabel("Bulan")
plt.ylabel("Total Number of Doctors Consultancy")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
